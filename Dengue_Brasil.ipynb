```python
def convertir_edad_sinan(x):
    if pd.isna(x):
        return np.nan

    try:
        x = int(x)
    except:
        return np.nan

    s = str(x).zfill(4)
    unidad = int(s[0])
    valor = int(s[1:])

    # 1 = horas, 2 = días, 3 = meses, 4 = años
    if unidad == 1:
        edad = valor / (24 * 365)
    elif unidad == 2:
        edad = valor / 365
    elif unidad == 3:
        edad = valor / 12
    elif unidad == 4:
        edad = valor
    else:
        return np.nan

    # Filtrar edades imposibles
    if edad < 0 or edad > 120:
        return np.nan

    return edad

df_modelo["EDAD_ANIOS"] = df_modelo["NU_IDADE_N"].apply(convertir_edad_sinan)
```

```python
# Calcular los cuartiles
Q1 = df_modelo["EDAD_ANIOS"].quantile(0.25)
Q3 = df_modelo["EDAD_ANIOS"].quantile(0.75)

# Calcular el rango intercuartílico
IQR = Q3 - Q1

# Calcular los límites
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)

# Identificar la cantidad de outliers antes
condicion_outlier = (
    (df_modelo["EDAD_ANIOS"] < limite_inferior) |
    (df_modelo["EDAD_ANIOS"] > limite_superior)
)

print("Outliers antes:", condicion_outlier.sum())

# Reemplazar los valores menores por el límite inferior
# y los mayores por el límite superior
df_modelo["EDAD_ANIOS"] = df_modelo["EDAD_ANIOS"].clip(
    lower=max(0, limite_inferior),
    upper=limite_superior
)

print("Nueva edad mínima:", df_modelo["EDAD_ANIOS"].min())
print("Nueva edad máxima:", df_modelo["EDAD_ANIOS"].max())
```

```python
# Recalcular cuartiles después del reemplazo
Q1_nuevo = df_modelo["EDAD_ANIOS"].quantile(0.25)
Q3_nuevo = df_modelo["EDAD_ANIOS"].quantile(0.75)

IQR_nuevo = Q3_nuevo - Q1_nuevo

limite_inferior_nuevo = Q1_nuevo - 1.5 * IQR_nuevo
limite_superior_nuevo = Q3_nuevo + 1.5 * IQR_nuevo

outliers_despues = (
    (df_modelo["EDAD_ANIOS"] < limite_inferior_nuevo) |
    (df_modelo["EDAD_ANIOS"] > limite_superior_nuevo)
).sum()

print("Outliers después:", outliers_despues)
print("Edad mínima final:", df_modelo["EDAD_ANIOS"].min())
print("Edad máxima final:", df_modelo["EDAD_ANIOS"].max())
```

##datos balanceados de Logistica

```python
def crear_pipeline_logistica_balanceado(
    pipeline_original,
    balanceador,
    class_weight=None
):
    """
    Construye un pipeline de Regresión Logística
    con una estrategia de balanceo.
    """

    # Copiar todos los pasos anteriores al modelo:
    # imputación, escalado, etc.
    pasos_preprocesamiento = [
        (nombre, clone(paso))
        for nombre, paso in pipeline_original.steps[:-1]
    ]

    # Copiar el modelo de Regresión Logística
    modelo = clone(
        pipeline_original.named_steps["modelo"]
    )

    # Configurar los pesos de clase
    modelo.set_params(
        class_weight=class_weight
    )

    # Crear pipeline de imbalanced-learn
    pipeline_balanceado = ImbPipeline(
        steps=pasos_preprocesamiento + [
            ("balanceo", balanceador),
            ("modelo", modelo)
        ]
    )

    return pipeline_balanceado
```

```python
def crear_pipeline_arbol_balanceado(
    pipeline_original,
    balanceador=None,
    class_weight=None
):
    """
    Construye un pipeline de Árbol de Decisión
    con SMOTE, Random UnderSampling o pesos de clase.
    """

    # Verificar que el último paso sea el modelo
    if pipeline_original.steps[-1][0] != "modelo":
        raise ValueError(
            "El último paso del pipeline debe llamarse 'modelo'."
        )

    # Copiar imputación y demás pasos anteriores al modelo
    pasos = [
        (nombre, clone(transformador))
        for nombre, transformador
        in pipeline_original.steps[:-1]
    ]

    # Copiar el Árbol de Decisión
    modelo_arbol = clone(
        pipeline_original.named_steps["modelo"]
    )

    # Aplicar o retirar los pesos algorítmicos
    modelo_arbol.set_params(
        class_weight=class_weight
    )

    # Añadir el balanceador solamente para SMOTE o UnderSampling
    if balanceador is not None:
        pasos.append(
            ("balanceo", clone(balanceador))
        )

    # Añadir el modelo al final
    pasos.append(
        ("modelo", modelo_arbol)
    )

    return ImbPipeline(steps=pasos)
```

```python
def crear_pipeline_rf_balanceado(
    pipeline_original,
    balanceador=None
):
    """
    Construye un pipeline de Random Forest con
    SMOTE, Random UnderSampling o pesos algorítmicos.
    """

    # Copiar todos los pasos anteriores al modelo
    pasos = [
        (nombre, clone(transformador))
        for nombre, transformador
        in pipeline_original.steps[:-1]
    ]

    # Copiar Random Forest
    modelo_rf = clone(
        pipeline_original.named_steps["modelo"]
    )

    # GridSearchCV manejará el paralelismo externo
    modelo_rf.set_params(
        n_jobs=1
    )

    # Agregar el método de balanceo
    if balanceador is not None:
        pasos.append(
            (
                "balanceo",
                clone(balanceador)
            )
        )

    # Agregar Random Forest al final
    pasos.append(
        (
            "modelo",
            modelo_rf
        )
    )

    return ImbPipeline(
        steps=pasos
    )
```

```python
class_weight=None
```

```python
def preparar_parametros_rf(
    parametros_originales,
    usar_pesos
):
    """
    Añade class_weight a la cuadrícula de búsqueda
    dependiendo de la estrategia utilizada.
    """

    parametros = copy.deepcopy(
        parametros_originales
    )

    # Convertir a lista para manejar diccionario
    # o lista de diccionarios
    es_diccionario = isinstance(
        parametros,
        dict
    )

    if es_diccionario:
        lista_parametros = [parametros]
    else:
        lista_parametros = parametros

    for cuadrícula in lista_parametros:

        if usar_pesos:
            cuadrícula[
                "modelo__class_weight"
            ] = [
                "balanced",
                "balanced_subsample"
            ]

        else:
            cuadrícula[
                "modelo__class_weight"
            ] = [None]

    if es_diccionario:
        return lista_parametros[0]

    return lista_parametros
```

```python
def crear_pipeline_sgd_balanceado(
    pipeline_original,
    balanceador=None
):
    """
    Construye un pipeline para SGD Classifier
    con SMOTE, Random UnderSampling o pesos.
    """

    # Verificar el nombre del último paso
    if pipeline_original.steps[-1][0] != "modelo":
        raise ValueError(
            "El último paso del pipeline debe llamarse 'modelo'."
        )

    # Copiar los pasos previos:
    # imputador, escalador, etc.
    pasos = [
        (nombre, clone(transformador))
        for nombre, transformador
        in pipeline_original.steps[:-1]
    ]

    # Añadir el balanceador cuando corresponda
    if balanceador is not None:
        pasos.append(
            ("balanceo", clone(balanceador))
        )

    # Copiar SGD Classifier
    modelo_sgd = clone(
        pipeline_original.named_steps["modelo"]
    )

    pasos.append(
        ("modelo", modelo_sgd)
    )

    return ImbPipeline(steps=pasos)
```

```python
def preparar_parametros_sgd(
    parametros_originales,
    class_weight
):
    """
    Añade class_weight a la cuadrícula de búsqueda
    sin modificar la configuración original.
    """

    parametros = copy.deepcopy(
        parametros_originales
    )

    # GridSearchCV acepta un diccionario
    # o una lista de diccionarios
    es_diccionario = isinstance(
        parametros,
        dict
    )

    if es_diccionario:
        lista_parametros = [parametros]
    else:
        lista_parametros = parametros

    for cuadrícula in lista_parametros:
        cuadrícula["modelo__class_weight"] = [
            class_weight
        ]

    if es_diccionario:
        return lista_parametros[0]

    return lista_parametros
```

```python
def crear_pipeline_hist_balanceado(
    pipeline_original,
    balanceador=None
):
    """
    Construye un pipeline de Hist Gradient Boosting
    con SMOTE, Random UnderSampling o pesos de clase.
    """

    if pipeline_original.steps[-1][0] != "modelo":
        raise ValueError(
            "El último paso del pipeline debe llamarse 'modelo'."
        )

    # Copiar los pasos previos al modelo
    pasos = [
        (nombre, clone(transformador))
        for nombre, transformador
        in pipeline_original.steps[:-1]
    ]

    # Agregar balanceador solo para SMOTE o UnderSampling
    if balanceador is not None:
        pasos.append(
            ("balanceo", clone(balanceador))
        )

    # Copiar el modelo
    modelo_hist = clone(
        pipeline_original.named_steps["modelo"]
    )

    pasos.append(
        ("modelo", modelo_hist)
    )

    return ImbPipeline(steps=pasos)
```

```python
def preparar_parametros_hist(
    parametros_originales,
    class_weight
):
    """
    Agrega class_weight a la cuadrícula de búsqueda
    de Hist Gradient Boosting.
    """

    parametros = copy.deepcopy(
        parametros_originales
    )

    # GridSearchCV permite diccionario o lista de diccionarios
    es_diccionario = isinstance(
        parametros,
        dict
    )

    if es_diccionario:
        lista_parametros = [parametros]
    else:
        lista_parametros = parametros

    for cuadricula in lista_parametros:

        cuadricula["modelo__class_weight"] = [
            class_weight
        ]

    if es_diccionario:
        return lista_parametros[0]

    return lista_parametros
```

```python
print("Métricas configuradas:")
print(metricas_grid)

if not isinstance(metricas_grid, dict):
    raise TypeError(
        "metricas_grid debe ser un diccionario."
    )

if "f1" not in metricas_grid:
    raise ValueError(
        "metricas_grid debe contener la métrica 'f1'."
    )
```

```python
metricas_grid = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1"
}
```

```python
from sklearn.model_selection import ParameterGrid

cantidad_combinaciones = len(
    list(ParameterGrid(parametros_hist_base))
)

cantidad_folds = validacion_cruzada.get_n_splits()

print("Combinaciones por estrategia:", cantidad_combinaciones)
print("Particiones de validación:", cantidad_folds)

print(
    "Entrenamientos por estrategia:",
    cantidad_combinaciones * cantidad_folds
)

print(
    "Entrenamientos para las tres estrategias:",
    cantidad_combinaciones * cantidad_folds * 3
)
```

```python
resultados_hist_balanceado = []
mejores_hist_balanceados = {}

# Limitar procesos para controlar el uso de memoria
N_JOBS_GRID = 2
```

```python
for nombre_estrategia, configuracion in estrategias_hist.items():

    print("\n" + "=" * 80)
    print("MODELO: HIST GRADIENT BOOSTING")
    print(f"ESTRATEGIA: {nombre_estrategia}")
    print("=" * 80)

    # Crear pipeline para la estrategia actual
    pipeline_actual = crear_pipeline_hist_balanceado(
        pipeline_original=pipeline_hist_base,
        balanceador=configuracion["balanceador"]
    )

    # Preparar hiperparámetros
    parametros_actuales = preparar_parametros_hist(
        parametros_originales=parametros_hist_base,
        class_weight=configuracion["class_weight"]
    )

    print("\nPipeline utilizado:")
    print(pipeline_actual)

    print("\nHiperparámetros evaluados:")
    print(parametros_actuales)

    # Configurar GridSearchCV
    grid_search_hist = GridSearchCV(
        estimator=pipeline_actual,
        param_grid=parametros_actuales,
        scoring=metricas_grid,
        refit="f1",
        cv=validacion_cruzada,
        n_jobs=N_JOBS_GRID,
        verbose=2,
        return_train_score=False,
        error_score="raise",
        pre_dispatch=N_JOBS_GRID
    )

    # Iniciar medición del tiempo
    inicio = time.perf_counter()

    # Entrenamiento
    grid_search_hist.fit(
        X_train,
        y_train
    )

    tiempo_entrenamiento = (
        time.perf_counter() - inicio
    )

    # Mejor pipeline
    mejor_hist = (
        grid_search_hist.best_estimator_
    )

    mejores_hist_balanceados[
        nombre_estrategia
    ] = mejor_hist

    # Predicción sobre el conjunto de prueba original
    y_pred = mejor_hist.predict(
        X_test
    )

    # Probabilidad de la clase hospitalizada
    y_prob = mejor_hist.predict_proba(
        X_test
    )[:, 1]

    # Matriz de confusión
    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    # Especificidad
    especificidad = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0
    )

    # Guardar métricas
    resultado = {
        "Modelo": "Hist Gradient Boosting",
        "Balanceo": nombre_estrategia,

        "Mejor_F1_CV":
            grid_search_hist.best_score_,

        "Accuracy_test":
            accuracy_score(
                y_test,
                y_pred
            ),

        "Balanced_accuracy_test":
            balanced_accuracy_score(
                y_test,
                y_pred
            ),

        "Precision_test":
            precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

        "Recall_test":
            recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

        "Especificidad_test":
            especificidad,

        "F1_test":
            f1_score(
                y_test,
                y_pred,
                zero_division=0
            ),

        "ROC_AUC_test":
            roc_auc_score(
                y_test,
                y_prob
            ),

        "PR_AUC_test":
            average_precision_score(
                y_test,
                y_prob
            ),

        "Verdaderos_negativos": tn,
        "Falsos_positivos": fp,
        "Falsos_negativos": fn,
        "Verdaderos_positivos": tp,

        "Tiempo_segundos":
            tiempo_entrenamiento,

        "Mejores_parametros":
            grid_search_hist.best_params_
    }

    resultados_hist_balanceado.append(
        resultado
    )

    print("\nRESULTADOS")
    print("-" * 60)

    print(
        f"Mejor F1 en validación cruzada: "
        f"{resultado['Mejor_F1_CV']:.4f}"
    )

    print(
        f"Accuracy en prueba: "
        f"{resultado['Accuracy_test']:.4f}"
    )

    print(
        f"Balanced Accuracy: "
        f"{resultado['Balanced_accuracy_test']:.4f}"
    )

    print(
        f"Precision de hospitalizados: "
        f"{resultado['Precision_test']:.4f}"
    )

    print(
        f"Recall de hospitalizados: "
        f"{resultado['Recall_test']:.4f}"
    )

    print(
        f"F1 de hospitalizados: "
        f"{resultado['F1_test']:.4f}"
    )

    print(
        f"Especificidad: "
        f"{resultado['Especificidad_test']:.4f}"
    )

    print(
        f"ROC-AUC: "
        f"{resultado['ROC_AUC_test']:.4f}"
    )

    print(
        f"PR-AUC: "
        f"{resultado['PR_AUC_test']:.4f}"
    )

    print(
        f"Tiempo: "
        f"{resultado['Tiempo_segundos']:.2f} segundos"
    )

    print(
        "Mejores hiperparámetros:",
        resultado["Mejores_parametros"]
    )

    # Guardar progreso después de cada estrategia
    pd.DataFrame(
        resultados_hist_balanceado
    ).to_csv(
        "resultados_hist_gradient_balanceado_progreso.csv",
        index=False,
        encoding="utf-8-sig"
    )
```
